```python
import pandas as pd
import os
```

```python
df = pd.read_csv('multiple_courses_modules2.csv')
```

```python
# Define the output directory
output_dir = 'courses_separated'
os.makedirs(output_dir, exist_ok=True)

# Group by the 'course' column to handle courses with multiple rows
grouped_courses = df.groupby('Course')

# Iterate over each course group
for course_name, group in grouped_courses:
    modules = []

    # For each group (course), combine modules from year 1 to year 4
    for year in ['Level 1', 'Level 3', 'Level 2', 'Level 4']:
        if year in df.columns:
            for module in group[year].dropna():  # Drop NaN values
                modules.extend(module.split(','))  # Split modules by commas if they are comma-separated

    # Remove any extra spaces
    modules = [module.strip() for module in modules if module.strip()]  # Also remove empty strings

    # If there are any modules collected, save them
    if modules:
        # Create a new DataFrame with the combined modules for the course
        modules_df = pd.DataFrame(modules, columns=['Modules'])

        # Save each course's modules to a CSV file, named after the course
        course_file_name = f"{course_name.replace(' ', '_')}_modules.csv"
        modules_df.to_csv(os.path.join(output_dir, course_file_name), index=False)

        # Display the DataFrame in Jupyter Notebook for verification
        print(f"Modules for {course_name}:")
        display(modules_df)  # Use display in Jupyter Notebook to see the DataFrame
```
